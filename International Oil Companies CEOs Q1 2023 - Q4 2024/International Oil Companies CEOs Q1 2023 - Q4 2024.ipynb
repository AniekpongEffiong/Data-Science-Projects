```python
# importing all modules
from sjvisualizer import DataHandler
from sjvisualizer import Canvas
from sjvisualizer import StackedBarChart
from sjvisualizer import ColumnRace
from sjvisualizer import StaticText
from sjvisualizer import Date
from sjvisualizer import ColumnRaceLegend, StackedRaceLegend
from sjvisualizer import StaticImage
import json
```

```python
# define constants
FPS = 60
DURATION = 0.25
BACKGROUND_COLOR = (247, 247, 247)
TOTAL_COLOR = (6,37,93)
FONT_COLOR = (14,45,101)
BAR_COLOR = (255,255,255)
STACK_COLOR = (250,250,250)
DATE_COLOR = (0, 0, 0)
GREY_COLOR = (163, 167, 181)
number_of_frames=FPS*DURATION*60
```

```python
# load the data from the excel file
df = DataHandler.DataHandler(excel_file="data/Earnings.xlsx", number_of_frames=number_of_frames).df
```

```python
# loading colors
colors = {
    "Woods": (254,0,12),
    "Sawan": (254,212,0),
    "Pouyanne": (249,249,249),
    "Wirth": (225,232,250),
}
```

```python
# creating the canvas to which we want to draw all elements
canvas = Canvas.canvas(bg=BACKGROUND_COLOR)
```

```python
# drawing a rectangle around working area
canvas.canvas.create_rectangle(-1,-1,1771,1771)
```

```python
# creating the bar race
column_race = ColumnRace.column_race(canvas=canvas.canvas, df=df, colors=colors, width=970, height=400, x_pos=685, y_pos=380, unit="$B", decimal_places = 0, font_color=DATE_COLOR, number_of_bars=5)
canvas.add_sub_plot(column_race)
```

```python
legend = StackedRaceLegend.legend(canvas=canvas.canvas, df=df, colors=colors, width=900, height=160, x_pos=990, y_pos=180, n=4, font_size=13, font_color=DATE_COLOR)
canvas.add_sub_plot(legend)
```

```python
# adding the time indication
time = Date.date(df=df, canvas=canvas.canvas, time_indicator="month", x_pos=690, y_pos=225, height=40, width=0, font_color=DATE_COLOR)
canvas.add_sub_plot(time)
```

```python
# adding a layer bar to the canvas
img = StaticImage.static_image(canvas=canvas.canvas, file="assets/Earnings.png",x_pos=405, y_pos=5, height=1070, width=1100,root=canvas.tk,on_top=False)
canvas.add_sub_plot(img)
```

```python
# starting the animation
canvas.play(fps=FPS)
```
